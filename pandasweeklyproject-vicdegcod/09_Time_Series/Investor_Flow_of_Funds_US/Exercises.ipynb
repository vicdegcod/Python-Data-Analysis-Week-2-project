```python
import pandas as pd

url = "https://raw.githubusercontent.com/datasets/investor-flow-of-funds-us/master/data/weekly.csv"

df = pd.read_csv(url)

df.head()
```

### Step 4.  What is the frequency of the dataset?

```python

# Convert the Date column to datetime
df["Date"] = pd.to_datetime(df["Date"])

# Set Date as the index
df.set_index("Date", inplace=True)

# Infer the frequency
print(pd.infer_freq(df.index))
```

### Step 8.  Change the frequency to monthly, sum the values and assign it to monthly.

```python
monthly = df.resample("ME").sum()
monthly
```

### Step 9. You will notice that it filled the dataFrame with months that don't have any data with NaN. Let's drop these rows.

```python
# Drop months with no data
monthly = monthly.dropna()

# Display the first few rows
print(monthly.head())
```

### Step 10. Good, now we have the monthly data. Now change the frequency to year.

```python
# Change the frequency to year-end and sum the values
yearly = df.resample("YE").sum()

# Display the first few rows
yearly.head()
```

### BONUS: Create your own question and answer it.

11. Which year recorded the highest total investor flow?

```python
# Calculate the total flow for each year
yearly["Total Flow"] = yearly.sum(axis=1)

# Find the year with the highest total flow
max_year = yearly["Total Flow"].idxmax()
max_flow = yearly["Total Flow"].max()

print("Year with the highest total investor flow:", max_year.year)
print("Total investor flow:", max_flow)
```
