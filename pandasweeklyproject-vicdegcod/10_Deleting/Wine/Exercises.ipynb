### Step 1. Import the necessary libraries

```python
import pandas as pd
import matplotlib.pyplot as plt
```

### Step 2. Import the dataset from this [address](https://archive.ics.uci.edu/ml/machine-learning-databases/wine/wine.data). 

### Step 3. Assign it to a variable called wine

```python
# Dataset URL
url = "https://archive.ics.uci.edu/ml/machine-learning-databases/wine/wine.data"

# Import the dataset
wine = pd.read_csv(url, header=None)

# Display the first five rows
print(wine.head())
```

### Step 4. Delete the first, fourth, seventh, nineth, eleventh, thirteenth and fourteenth columns

```python
# Delete the specified columns
wine = wine.drop(wine.columns[[0, 3, 6, 8, 10, 12, 13]], axis=1)

# Display the first five rows
print(wine.head())
```

### Step 5. Assign the columns as below:

The attributes are (donated by Riccardo Leardi, riclea '@' anchem.unige.it):  
1) alcohol  
2) malic_acid  
3) alcalinity_of_ash  
4) magnesium  
5) flavanoids  
6) proanthocyanins  
7) hue 

```python
# Assign new column names
wine.columns = [
    "alcohol",
    "malic_acid",
    "alcalinity_of_ash",
    "magnesium",
    "flavanoids",
    "proanthocyanins",
    "hue"
]

# Display the first five rows
print(wine.head())
```

### Step 6. Set the values of the first 3 rows from alcohol as NaN

```python
import numpy as np
# Set the first 3 rows of the 'alcohol' column to NaN
wine.loc[0:2, 'alcohol'] = np.nan

# Display the first five rows to verify
print(wine.head())
```

### Step 7. Now set the value of the rows 3 and 4 of magnesium as NaN

```python
import numpy as np

# Set rows 3 and 4 of the 'magnesium' column to NaN
wine.loc[3:4, 'magnesium'] = np.nan

# Display the first six rows to verify
print(wine.head(6))
```

### Step 8. Fill the value of NaN with the number 10 in alcohol and 100 in magnesium

```python
# Replace NaN values
wine['alcohol'] = wine['alcohol'].fillna(10)
wine['magnesium'] = wine['magnesium'].fillna(100)

# Display the first six rows to verify
print(wine.head(6))
```

### Step 9. Count the number of missing values

```python
# Count missing values in each column
print(wine.isnull().sum())
```

### Step 12.  How many missing values do we have?

```python
print(wine.isnull().sum())
```

### Step 13. Delete the rows that contain missing values

```python
# Delete rows that contain any NaN values
wine = wine.dropna()

# Display the first five rows
print(wine.head())
```

### Step 14. Print only the non-null values in alcohol

```python
# Print only the non-null values in the 'alcohol' column
print(wine['alcohol'].dropna())
```

### Step 15.  Reset the index, so it starts with 0 again

```python
# Reset the index
wine.reset_index(drop=True, inplace=True)

# Display the first five rows
print(wine.head())
```
