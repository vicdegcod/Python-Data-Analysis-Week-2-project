### Step 4. Create columns for the dataset

```python
# Dataset URL
url = "https://archive.ics.uci.edu/ml/machine-learning-databases/iris/iris.data"

# Column names
columns = [
    "sepal_length",
    "sepal_width",
    "petal_length",
    "petal_width",
    "class"
]

# Import the dataset with the column names
iris = pd.read_csv(url, header=None, names=columns)

# Display the first five rows
print(iris.head())
```

### Step 5.  Is there any missing value in the dataframe?

```python
# Check for missing values in each column
print(iris.isnull().sum())
```

### Step 6.  Lets set the values of the rows 10 to 29 of the column 'petal_length' to NaN

```python
import numpy as np

# Set rows 10 to 29 of 'petal_length' to NaN
iris.loc[10:29, 'petal_length'] = np.nan

# Verify the changes
print(iris.loc[10:29])
```

### Step 7. Good, now lets substitute the NaN values to 1.0

```python
# Replace NaN values in the 'petal_length' column with 1.0
iris['petal_length'] = iris['petal_length'].fillna(1.0)

# Verify that there are no missing values left
print(iris['petal_length'].isnull().sum())
```

### Step 8. Now let's delete the column class

```python
# Delete the 'class' column
iris = iris.drop(columns=['class'])

# Display the first five rows
print(iris.head())
```

### Step 9.  Set the first 3 rows as NaN

```python
import numpy as np

# Set the first 3 rows to NaN
iris.iloc[0:3] = np.nan

# Display the first five rows to verify
print(iris.head())
```

### Step 10.  Delete the rows that have NaN

```python
# Delete rows that contain NaN values
iris = iris.dropna()

# Display the first five rows
print(iris.head())
```

### Step 11. Reset the index so it begins with 0 again

```python
# Reset the index
iris.reset_index(drop=True, inplace=True)

# Display the first five rows
print(iris.head())
```
